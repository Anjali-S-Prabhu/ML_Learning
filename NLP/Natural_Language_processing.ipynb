# WAP to display other information about each text by looping over all the values of fileid corresponding to the gutenberg file

```python
for fileid in gutenberg.fileids():
    num_chars=len(gutenberg.raw(fileid))
    num_words=len(gutenberg.words(fileid))
    num_sents=len(gutenberg.sents(fileid))
    num_vocab=len(set([w.lower() for w in gutenberg.words(fileid)]))
    print(int(num_chars/num_words),int(num_words/num_sents),int(num_words/num_vocab),fileid)
```

```python
from nltk.tokenize import word_tokenize,sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer,WordNetLemmatizer
from nltk import pos_tag
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from collections import Counter
import pandas as pd
```

```python
text = """In December 1928, Bhagat Singh and an associate, Shivaram Rajguru, both members of a small revolutionary group, the Hindustan Socialist Republican Association (also Army, or HSRA), shot dead a 21-year-old British police officer, John P. Saunders, in Lahore, Punjab, in what is today Pakistan, mistaking Saunders, who was still on probation, for the British senior police superintendent, James Scott, whom they had intended to assassinate. They held Scott responsible for the death of a popular Indian nationalist leader Lala Lajpat Rai for having ordered a lathi (baton) charge in which Rai was injured and two weeks thereafter died of a heart attack. As Saunders exited a police station on a motorcycle, he was felled by a single bullet fired from across the street by Rajguru, a marksman."""
```

```python
sentences=sent_tokenize(text)
print("\n"+"="*60)
print("1.sentence tokenization")
print("="*60)
for i ,s in enumerate(sentences,1):
    print(f"Sentence{i}: {s}")
```

```python
words=word_tokenize(text)
print("\n"+"="*60)
print("2.word tokenization")
print("="*60)
print("Tokens:",words)
print('Totla token',len(words))
```

```python
# cleaning(Lower case remove punctuation & numbers)
cleaned_words=[w.lower()for w in words if w.isalpha()]
print("\n"+"="*60)
print(cleaned_words)
```

```python
# remove the stopwords
stop_words=set(stopwords.words('english'))
filtered_word=[w for w in cleaned_words if w not in stop_words]
print("\n"+"="*60)
print("Filtered_word")
print(filtered_word)
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_word]

print("\n" + "=" * 60)
print("Stemming (Porter Stemmer)")
print("=" * 60)

for original, stemmed in zip(filtered_word, stemmed_words):
    print(f"{original:15} -> {stemmed}")
```

## Lemmatizer

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

lemmatized_word = [lemmatizer.lemmatize(w) for w in filtered_word]

print('\n' + "=" * 60)
print("Lemmatization of word")
print("=" * 60)

for orig, lemma in zip(filtered_word, lemmatized_word):
    print(f"{orig:15} -> {lemma}")
```

```python
def lemmatize_with_pos(word,tag):
    """map pos tag to wordnet format and lemmatize"""
    tag_map={'j':'a','N':'n','V':'v','R':'r'}
    pos=tag_map.get(tag[0],'n')
    return lemmatizer.lemmatize(word,pos=pos)
```

```python
pos_tagged = pos_tag(filtered_words)

lemmatized_pos = [
    lemmatize_with_pos(w, t)
    for w, t in pos_tagged
]

print(lemmatized_pos)
```
