```python
import pandas as pd
import numpy as np
```

```python
data={
    "student":["amit","rahul","priya","sneha","karan","neha","rohit","pooja","arjun","simran"],
    "age":[20,21,np.nan,22,20,23,np.nan,21,22,20],
    "marks":[85,90,78,np.nan,88,92,76,81,np.nan,89],
    "city":["delhi","mumbai","delhi","kolkata","np.nan","mumbai","delhi","chennai","kolkata",np.nan]
}
df=pd.DataFrame(data)
print(df)
```
