```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
np.random.seed(42)
x=np.linspace(0,10,20).reshape(-1,1)
y=2*x.squeeze()+3+np.random.normal(0,2,20)
print(x)
print(y)
print(x.shape)
```

```python
model = LinearRegression()
model.fit(x, y)
```

```python
def predict(x, w, b):
    return w * x + b
```

```python
model = LinearRegression()
model.fit(x, y)

y_pred = model.predict(x)

plt.figure(figsize=(8,6))

plt.scatter(x, y, color='red', label="Training data")

plt.plot(x, y_pred, color='blue', linewidth=3, label="Linear Regression")

plt.title("Simple Linear Regression Plotting")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.grid()

plt.show()
```

```python
degrees=[2,5,10,15]
x_test=np.linspace(0,10,300).reshape(-1,1)
fig,axes=plt.subplots(nrows=2,ncols=2,figsize=(14,14))
axes=axes.ravel()

for i, degree in enumerate(degrees):
    poly=PolynomialFeatures(degree=degree)
    model=LinearRegression()
    x_poly=poly.fit_transform(x)
    model.fit(x_poly,y)
    x_test_poly=poly.transform(x_test)
    y_pred=model.predict(x_test_poly)
    axes[i].scatter(x,y,color="red",label="training data")
    axes[i].plot(x_test,y_pred,color='blue',linewidth=3,label=f'degree{degree}')
    axes[i].set_title(f'plonomail Degree={degree}',fontsize=14)
    axes[i].legend()

plt.show()
```
