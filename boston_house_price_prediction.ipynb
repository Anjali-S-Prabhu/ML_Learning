1)Qualitative(categorical) DATA

## Data that describe qualities or characteristics, and cannot be measured numerically, that data is know as categorical or qualitative dada

a) Nominal data :- categories with no inherent order

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
nominal_data=pd.Series(['Red','Blue','Green','Red','Blue','Red','Green','Blue','Red','Green','Blue'],name='Colors')
print(nominal_data)
print(nominal_data.value_counts())
```

```python
nominal_data.value_counts().plot(kind='bar',color='skyblue')
plt.title("Nominal data - colors")
plt.xlabel('color')
plt.ylabel('frequency')
plt.show()
```

## orfinal data:-categories with a meaning full order/ranking

```python
import pandas as pd

ordinal_data = pd.Series(
    ['Low', 'Medium', 'High', 'Medium', 'Low',
     'High', 'High', 'Medium', 'Low', 'Medium', 'High'],
    name='Satisfaction'
)

order = ['Low', 'Medium', 'High']

ordinal_data = pd.Categorical(
    ordinal_data,
    categories=order,
    ordered=True
)

print(ordinal_data.value_counts().sort_index())
```

```python
ordinal_data.value_counts().sort_index().plot(kind='bar',color='lightgreen')
plt.title("Ordinal_data - Satisfaction level")
plt.show()
```

## normal distribution : a distribution shows how data values are spread across a range, the KDE is a sooth curve that estimate the probability of the data like a smoothes histogram

## <font color='red'> The three key distribution are,

1)normal (symmetric) (bell shaped) balanced both side

2)right skewed (positive) long tall tail on right side mean>median

3)left skewed (negative) long tail on left side mean<median

```python
import numpy as np
import pandas as pd

np.random.seed(42)

n = 10000

normal_data = np.random.normal(loc=50, scale=10, size=n)
right_skewed = np.random.exponential(scale=10, size=n) + 20
left_skewed = 100 - np.random.exponential(scale=10, size=n)

df = pd.DataFrame({
    'Normal': normal_data,
    'Right_skewed': right_skewed,
    'Left_skewed': left_skewed
})

df.head()
```
