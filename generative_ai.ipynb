```python
# Reusable function
def ask_llama(question):
    response = ollama.chat(
        model="llama3.2",
        messages=[
            {
                'role': 'user',
                'content': question
            }
        ]
    )
    return response['message']['content']
```

```python
answer=ask_llama("what is nlp")
print(answer)
```

```python
def ai_teacher(question):
    response = ollama.chat(
        model='llama3.2',
        messages=[
            {
                'role': 'system',
                'content': """You are an expert college lecturer.
Teach students in a simple and practical way, explaining concepts
step by step. Use real-world examples and Python examples whenever
appropriate. Do not assume advanced knowledge."""
            },
            {
                'role': 'user',
                'content': question
            }
        ]
    )
    
    return response['message']['content']
```

```python
print(ai_teacher('explain neural network'))
```

```python
def python_expert(question):
    response=ollama.chat(
        model="llama3.2",
         messages=[
            {
                'role': 'user',
                'content': """ you are an expert python developer.provide clean python code."""
            },
            {
                'role': 'user',
                'content': question
            }
        ]
    )
    return response['message']['content']
```

```python
print(python_expert("write factorial recursion program"))
```

```python
conversation = []

def chat_with_memory(question):
    
    # Store user's question
    conversation.append({
        "role": "user",
        "content": question
    })
    
    # Send entire conversation to Ollama
    response = ollama.chat(
        model="llama3.2",
        messages=conversation
    )
    
    # Get AI answer
    answer = response["message"]["content"]
    
    # Store AI answer in conversation
    conversation.append({
        "role": "assistant",
        "content": answer
    })
    
    return answer
```

```python
print(chat_with_memory("my name is anjali"))
```

```python
print(chat_with_memory(" i am teaching a student about generative AI"))
```

```python
print(chat_with_memory("what is my name?"))
```

```python
print(chat_with_memory("what subject am teaching"))
```

```python
response = ollama.chat(
    model="llama3.2",
    messages=[
        {
            "role": "user",
            "content": "Explain Generative AI"
        }
    ],
    stream=True
)

for chunk in response:
    print(chunk["message"]["content"], end="", flush=True)
```

```python
def ask_llama_stream(question):
    response = ollama.chat(
        model="llama3.2",
        messages=[
            {
                "role": "user",
                "content": question
            }
        ],
        stream=True
    )

    for chunk in response:
        print(chunk["message"]["content"], end="", flush=True)

    print()
```

```python
ask_llama_stream("Explain the karnatka politics")
```
