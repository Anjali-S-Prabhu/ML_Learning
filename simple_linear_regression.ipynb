```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
x=2*np.random.rand(100,1)
y=4+3*x+np.random.randn(100,1)
```

```python
plt.figure(figsize=(12,8))
plt.scatter(x,y,color='blue',marker='o',label='scatter')
plt.xlabel("The x axis")
plt.ylabel("The y axis")
plt.show()
```

```python
# adding the bias
x_b=np.c_[np.ones((100,1)),x]
```

```python
Theta_best=np.linalg.inv(x_b.T.dot(x_b)).dot(x_b.T).dot(y)
```

```python
Theta_best
```

```python
x_new = np.array([[0], [2]])

x_new_b = np.c_[np.ones((2, 1)), x_new]

y_predict = x_new_b.dot(Theta_best)

print(y_predict)
```

```python
plt.figure(figsize=(12, 8))

plt.scatter(x, y, color='blue', marker='o', label='scatter')
plt.plot(x_new, y_predict, "r-", label='prediction')

plt.xlabel("The x axis")
plt.ylabel("The y axis")
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("boston.csv")

df.head()
```

```python
corr = df.corr(numeric_only=True)
corr_value = corr["medv"].sort_values(ascending=False)
corr_value
```

```python
correlated_data = df[corr_value.index]
correlated_data.head()
```

```python
correlated_data = df[corr_value.index]
correlated_data.head()
```

```python
sns.heatmap(correlated_data.corr(), annot=True, annot_kws={'size':10})
plt.show()
```

```python
x=correlated_data[['rm','ptratio','lstat']]
y=correlated_data['medv']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
lin_reg=LinearRegression()
```

```python
lin_reg.fit(x_train,y_train)
```

```python
y_pred=lin_reg.predict(x_test)
```

```python
result_df=pd.DataFrame(data=[y_pred,y_test],index=["predicted_value","actual_value"])
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
print("the mse is",mean_squared_error(y_test,y_pred))
print("the mean absolute error is",mean_absolute_error(y_test,y_pred))
print("the r2 score is",r2_score(y_test,y_pred)*100)
```
