```python
np.random.seed(0) #42
df1=pd.DataFrame(dict(time=np.arange(500),value=np.random.random(500).cumsum()))
df1.head()
```

```python
sns.relplot(x='time',y='value',data=df1,kind='line')
```

```python
df2=pd.DataFrame(np.random.randn(500,2).cumsum(axis=0),columns=['time','value'])
df2.head()
```

```python
sns.relplot(x='time',y='value',kind='line',data=df2,sort=True)
```

```python
sns.relplot(x='time',y='value',kind='line',data=df2,sort=False)
```
