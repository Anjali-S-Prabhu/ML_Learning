## write a  program that prompts the user prompts the user for two number and display the addition, subtraction,  multiplication, and division between them


```python
#prompt the user for two number
num1 = float(input("enter the first number"))
num2 = float(input("enter the second number"))

#perform arithmetic operation
addition = num1+num2
subtraction = num1-num2
multiplication = num1*num2
division = num1/num2

#display operation
print("addition is",addition)
print("subtraction is",subtraction)
print("multiplication is",multiplication)
print("division is",division)

```

## WAP that calculate the geometric mean of 3 numbers entered by the users

```python
import math

num1 = float(input("enter the first number"))
num2 = float(input("enter the second number"))
num3 = float(input("enter the third number"))
product = num1*num2*num3
g_mean = math.pow(product,1/3)
print("the value of geometric mean is :",g_mean)
```

## WAP that prints all even number from 1 to 100

```python
num = 1
while num<=100:
    if num % 2 == 0:
        print(num)
    num = num+1
```

## WAP that display even numbers 1 to 50 and odd number 51 to 100 using repeating loop

```python
num = 1
while num<=100:
    if num<=50:
        if num%2==0:
            print("even:",num)
    else:
        if num%2!=0:
            print("odd:",num)
            
    num = num+1
```

## WAP to display and find the sum of a list numbers using for loop

```python
l = [10,20,30,40,50]
s = 0
for i in l:
    print(i)
    s=s+i
print("sum of list is",s)
```
